```python
SELECT 
    eo.year,
    sum(eo.total_profit) AS total_profit
FROM pem_curated.enriched_orders as eo
GROUP BY 1
ORDER BY 1 desc;
```

```python
SELECT 
    eo.year,
    eo.category,
    sum(eo.total_profit) AS total_profit
FROM pem_curated.enriched_orders as eo
GROUP BY 1,2
ORDER BY 1 desc;
```

```python
SELECT 
    eo.customer_name,
    sum(eo.total_profit) AS total_profit
FROM pem_curated.enriched_orders as eo
GROUP BY 1
ORDER BY 2 DESC;
```
